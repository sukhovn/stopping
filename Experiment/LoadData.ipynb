```python
from pymongo import MongoClient
import pymongo
import numpy as np
import pandas as pd

#Loading mongodb access link
file = open("db_user_pass.txt",mode='r')
CONNECTION_STRING = file.read()
if CONNECTION_STRING.endswith('\n'):
    CONNECTION_STRING = CONNECTION_STRING[:-1]

#loading 
client = MongoClient(CONNECTION_STRING)
entries_collection = client['button_experiment']['pilot_2']

#putting database in pandas
df = pd.DataFrame(entries_collection.find()).drop(columns=['_id','__v'])
```

```python
#selecting subjects from prolific only
subjects = df[1:].reset_index(drop=True)
subjects.to_pickle('pilot_2_results.pkl')
#saving prolificIDs in a csv file
subject_wins = subjects[['prolificID']]
wins_to_dollars = (1.8 + subjects['total_win']*0.01).mul(100).apply(np.ceil).mul(0.01).round(2)
subject_wins.insert(1, 'Win', wins_to_dollars)
subject_wins.to_csv("prolific_wins.csv")

#selecting subjects, 1 and 54 are my two friends, 4:54 are the actual subjects
# subjects_full = df.iloc[[1] + [*range(4,54)] + [54]].drop(columns=['prolificID']).reset_index(drop=True)
# subjects_full.to_pickle("pilot_results_full.pkl")
```
